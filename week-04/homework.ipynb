**Review Problem 1**

**Concept:** *Functions with lists and indexing.*

**Task:**

Write a function middlest_element(values) that returns the middle element of a list.

Rules:

* If the list is empty, return `None`.

* If the list length is odd, return the element at the exact middle index.

* If the list length is even, return the element just left of center (example: for length 6, use index 2).

* Then run the provided test cases and print results exactly as shown.

**Sample Output:**

```
middlest_element([1,2,3,4,5]) =  3
middlest_element([1,2,3,4,5,6]) =  3
middlest_element([]) =  None
```


````python
def middlest_element(values):
    """Return the middle element of a list.

    For even-length lists, return the element just left of center.
    For empty lists, return None.
    """
    if len(values) == 0:
        return None

    middle_index = (len(values) - 1) // 2
    return values[middle_index]
````

Rules:

````python
def square_evens(values):
    """Return the squares of all even numbers in values."""
    return [n * n for n in values if n % 2 == 0]
````

````python
def print_names(names):
    """Print each name on its own line using an index-based loop."""
    for i in range(len(names)):
        print(names[i])


def print_adjacent_names(names):
    """Print each name with the next name on the same line."""
    for i in range(len(names) - 1):
        print(f'{names[i]} , {names[i + 1]}')
````

**Problem 6**

**Concept:** *Accumulator pattern with numeric result, list iteration.*

**Task:**

Write a function `mean(values)` that returns the arithmetic mean of a list of numbers.

Rules:

* Must use a loop + accumulator (no `sum()`).

* No imports.

* Then run the given test case and print exactly in the same format.

**Sample Output:**

```

```


````python
def mean(values):
    """Return the arithmetic mean of a list of numbers."""
    total = 0
    for value in values:
        total += value
    return total / len(values)
````

**Problem 7**

**Concept:** *Accumulator pattern with numeric result, list iteration.*

**Task:**

Write a function `variance(values)` that returns the population variance of a list of numbers using this formula:

$$
\text{variance} = \frac{1}{n} \sum_{i=1}^{n} (x_i - \bar{x})^2
$$

Rules:

* Must use loops + accumulator (no `sum()`).

* No imports.

* Must call your `mean(values)` function to get the mean.

* Then run the provided test case and print in the same format.

**Sample Output:**

```
values: [6, 3, 7, 9, 5]
mean: 6.0
variance: 4.0
```


````python
def variance(values):
    """Return the population variance of a list of numbers."""
    avg = mean(values)
    total = 0
    for value in values:
        total += (value - avg) ** 2
    return total / len(values)
````

````python
values = [6, 3, 7, 9, 5]
print('values:', values)
print('mean:', mean(values))
print('variance:', variance(values))
````

**Problem 8**

**Concept:** *Accumulator pattern with numeric result, list iteration, index-based loop.*

**Task:**

Write a function `covariance(xs, ys)` that computes the population covariance between two series:

$$
\text{covariance} = \frac{1}{n} \sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})
$$

Rules:

* If `xs` and `ys` are not the same length, print:
    
    `xs and ys must be the same length`
    
    and return `0`.

* Must use an index-based loop.

* Must use your `mean()` function.

* Then run the given test case and print exactly in the same format.

**Sample Output:**

```
x values: [6, 3, 7, 9, 5]
y values: [4, 1, 7, 5, 0]
covariance: 3.8
```


````python
def covariance(xs, ys):
    """Return the population covariance of two equal-length lists."""
    if len(xs) != len(ys):
        print('xs and ys must be the same length')
        return 0

    x_mean = mean(xs)
    y_mean = mean(ys)
    total = 0
    for i in range(len(xs)):
        total += (xs[i] - x_mean) * (ys[i] - y_mean)
    return total / len(xs)
````

Test cases for Problem 8

````python
xs = [6, 3, 7, 9, 5]
ys = [4, 1, 7, 5, 0]
print('x values:', xs)
print('y values:', ys)
print('covariance:', covariance(xs, ys))
````

Write a function `polynomial_value(x, coefficients)` that evaluates a polynomial where each element in `coefficients` is the coefficient for

Example meaning:



````python
def polynomial_value(x, coefficients):
    """Evaluate a polynomial using coefficients[i] * x**i."""
    result = 0
    for i in range(len(coefficients)):
        result += coefficients[i] * (x ** i)
    return result
````

**Problem 12**

**Concept:** *Accumulator pattern to process a string, produce a string result.*

**Task:**

Write a function `cipher(s, n)` that applies a Caesar cipher shift to a plaintext string.

Rules:

* Assume lowercase letters only for shifting (`a` to `z`).

* Do not shift non-letters (spaces stay spaces, punctuation stays punctuation).

* Must wrap around the alphabet if shifting past `z` or before `a`.

* Must use an accumulator string.

* Then run the exact test cases shown and print in the same format.

**Sample Output:**

```
plaintext = abc, ciphered = def
plaintext = abc xyz, ciphered = def abc
```


````python
def cipher(s, n):
    """Apply a Caesar cipher to lowercase letters in s."""
    result = ''
    for ch in s:
        if 'a' <= ch <= 'z':
            offset = ord(ch) - ord('a')
            new_offset = (offset + n) % 26
            result += chr(ord('a') + new_offset)
        else:
            result += ch
    return result
````

Test case for problem 12.

````python
plaintext = 'abc'
result = cipher(plaintext, 3)
print(f'plaintext = {plaintext}, ciphered = {result}')

plaintext = 'abc xyz'
result = cipher(plaintext, 3)
print(f'plaintext = {plaintext}, ciphered = {result}')
````

Write a function `print_all_decipherings(s)` that prints all 26 possible Caesar shifts of string `s`.

Rules:

* Use a definite loop for shifts `0` through `25`.

* For each shift `i`, compute `ciphered = cipher(s, i)` and print:

    `{i}, {ciphered}`

* Then run these two test cases:

**Sample Output:**

````python
def print_all_decipherings(s):
    """Print every Caesar shift of s from 0 through 25."""
    for i in range(26):
        ciphered = cipher(s, i)
        print(f'{i}, {ciphered}')
````
