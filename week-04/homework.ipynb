```python
def middlest_element(values):
    """Return the middle element of a list.

    If the list is empty, return None. For even lengths, return the
    element just left of center.
    """
    if not values:
        return None

    middle = len(values) // 2
    if len(values) % 2 == 0:
        return values[middle - 1]
    return values[middle]

print('middlest_element([1,2,3,4,5]) = ', middlest_element([1, 2, 3, 4, 5]))
print('middlest_element([1,2,3,4,5,6]) = ', middlest_element([1, 2, 3, 4, 5, 6]))
print('middlest_element([]) = ', middlest_element([]))

values = [0, 1, 2, 3, 7, 8, 9]
print('middlest_element(values) = ', middlest_element(values))

values = [1, 2, 3, 7, 8, 9]
print('middlest_element(values) = ', middlest_element(values))

values = []
print('middlest_element(values) = ', middlest_element(values))
```

```python
def square_evens(values):
    """Return the square of each even number in the original order."""
    return [value * value for value in values if value % 2 == 0]

print(square_evens([3, 4, 2, 6, 9, 8]))
assert square_evens([3, 4, 2, 6, 9, 8]) == [16, 4, 36, 64]
assert square_evens([1, 3, 5, 7]) == []
assert square_evens([]) == []
assert square_evens([-2, -5, 4, -8]) == [4, 16]
print('All square_evens tests passed.')
```

```python
def print_powers(base, n):
    """Print the first n+1 powers of base from exponent 0 to n."""
    print(f'printing the first {n + 1} powers of {base}')
    for i in range(n + 1):
        print(f'{base} ** {i} = {base ** i}')

print_powers(2, 10)
```

```python
def print_even_odd(values):
    """Print whether each element in values is even or odd."""
    for value in values:
        if value % 2 == 0:
            print(f'{value} is even')
        else:
            print(f'{value} is odd')

values = [8, 7, 6, 5, 3, 0, 9]
print_even_odd(values)
```

```python
def print_names(names):
    """Print each name on its own line using an index-based loop."""
    for i in range(len(names)):
        print(names[i])

def print_adjacent_names(names):
    """Print each name paired with the next name in the list."""
    for i in range(len(names) - 1):
        print(f'{names[i]} , {names[i + 1]}')

names = ['John', 'Paul', 'George', 'Ringo', 'Billy']
print("calling print_names(names)")
print_names(names)
print("calling print_adjacent_names(names)")
print_adjacent_names(names)
```

```python
def mean(values):
    """Return the arithmetic mean of a list of numbers."""
    total = 0
    for value in values:
        total += value
    if len(values) == 0:
        return 0
    return total / len(values)

values = [2, 1, 6, 3, 7]
print(f'values: {values}')
print(f'mean: {mean(values)}')
```

```python
def variance(values):
    """Return the population variance of a list of numbers."""
    avg = mean(values)
    total = 0
    for value in values:
        total += (value - avg) ** 2
    if len(values) == 0:
        return 0
    return total / len(values)

values = [6, 3, 7, 9, 5]
print(f'values: {values}')
print(f'mean: {mean(values)}')
print(f'variance: {variance(values)}')
assert variance(values) == 4.0
```

```python
def covariance(xs, ys):
    """Return the population covariance of two equal-length lists."""
    if len(xs) != len(ys):
        print('xs and ys must be the same length')
        return 0

    x_mean = mean(xs)
    y_mean = mean(ys)
    total = 0
    for i in range(len(xs)):
        total += (xs[i] - x_mean) * (ys[i] - y_mean)
    return total / len(xs)

x_values = [6, 3, 7, 9, 5]
y_values = [4, 1, 7, 5, 0]
print(f'x values: {x_values}')
print(f'y values: {y_values}')
print(f'covariance: {covariance(x_values, y_values)}')
assert covariance(x_values, y_values) == 3.8
```

```python
def polynomial_value(x, coefficients):
    """Evaluate a polynomial with coefficients for powers of x from 0 upward."""
    result = 0
    for i in range(len(coefficients)):
        result += coefficients[i] * (x ** i)
    return result

coefficients = [3, 1, 2]
print(f'result = {polynomial_value(2, coefficients)}')
assert polynomial_value(2, coefficients) == 3 * (2 ** 0) + 1 * (2 ** 1) + 2 * (2 ** 2)
print(f'result = {polynomial_value(-1, [8, 6, 7, 5])}')
```

```python
def string_to_ascii(s):
    """Convert each character in s to its ASCII value."""
    result = []
    for character in s:
        result.append(ord(character))
    return result

result = string_to_ascii('hello')
print(f'result = {result}')
assert result == [104, 101, 108, 108, 111]
```

```python
def ascii_to_string(codes):
    """Convert a list of ASCII codes into the matching string."""
    result = ''
    for code in codes:
        result += chr(code)
    return result

codes = [121, 111, 117, 32, 103, 111, 116, 32, 105, 116, 33]
print(f'codes = {codes}')
print(ascii_to_string(codes))
assert ascii_to_string(codes) == 'you got it!'
```

```python
def cipher(s, n):
    """Apply a Caesar shift to lowercase letters in s."""
    result = ''
    for character in s:
        if 'a' <= character <= 'z':
            shifted = ord('a') + (ord(character) - ord('a') + n) % 26
            result += chr(shifted)
        else:
            result += character
    return result

plaintext = 'abc'
print(f'plaintext = {plaintext}, ciphered = {cipher(plaintext, 3)}')
assert cipher(plaintext, 3) == 'def'
plaintext = 'abc xyz'
print(f'plaintext = {plaintext}, ciphered = {cipher(plaintext, 3)}')
assert cipher(plaintext, 3) == 'def abc'
```

```python
def print_all_decipherings(s):
    """Print all 26 Caesar shifts of s."""
    for i in range(26):
        print(f'{i}, {cipher(s, i)}')

print_all_decipherings('def')
print_all_decipherings('oqfcgg hvs ibwjsfgs')
```
